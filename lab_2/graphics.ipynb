```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("timings.csv")

#только линейный поиск
linear_df = df[df["algorithm"] == "linearSearch"].sort_values("N")

plt.figure(figsize=(10, 6))
plt.plot(linear_df["N"], linear_df["time_us"], marker="o", label="linearSearch")
plt.xlabel("Размер массива N")
plt.ylabel("Время (микросекунды)")
plt.title("Линейный поиск")
plt.grid(True)
plt.legend()
plt.savefig("linear_search_times.png", dpi=200)
plt.show()

# все остальные алгоритмы
other_df = df[df["algorithm"] != "linearSearch"]

plt.figure(figsize=(10, 6))
for algo, sub in other_df.groupby("algorithm"):
    sub = sub.sort_values("N")
    plt.plot(sub["N"], sub["time_us"], marker="o", label=algo)

plt.xlabel("Размер массива N")
plt.ylabel("Время (микросекунды)")
plt.title("BST, RBTree, HashTable и multimap")
plt.grid(True)
plt.legend()
plt.savefig("other_search_times.png", dpi=200)
plt.show()
```

```python
df = pd.read_csv("collisions.csv")

plt.figure(figsize=(10, 6))
plt.plot(df["N"], df["count"], marker="o")
plt.xlabel("Размер массива N")
plt.ylabel("Количество коллизий")
plt.title("Зависимость количества коллизий от размера выборки")
plt.grid(True)
plt.legend()
plt.savefig("collisions.png", dpi=200)
plt.show()

plt.figure(figsize=(10, 6))
plt.plot(df["N"], df["ratio"], marker="o")
plt.xscale("log")
plt.xlabel("Размер массива N")
plt.ylabel("Относительная величина коллизий")
plt.title("Зависимость относительного количества коллизий от размера выборки")
plt.grid(True)
plt.legend()
plt.savefig("relative_collisions.png", dpi=200)
plt.show()
```
