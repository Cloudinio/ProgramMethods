```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("speed.csv")

plt.figure(figsize=(10, 6))

for method in df["method"].unique():
    method_df = df[df["method"] == method]
    plt.plot(
        method_df["N"],
        method_df["milliseconds"],
        marker="o",
        label=method
    )

plt.xlabel("Размер выборки N")
plt.ylabel("Время генерации, мс")
plt.title("Зависимость времени генерации от размера выборки")
plt.grid(True)
plt.legend()
plt.savefig("speed_by_sample_size.png", dpi=200)
plt.show()
```
